```python
import numpy as np
import matplotlib.pyplot as plt                                                                                                                                                                                 # Função de modulação
def modulator(values):
    raiz2 = np.sqrt(2)
    mapping = {
        0: (1/raiz2) + 1j*(1/raiz2),
        1: (-1/raiz2) + 1j*(1/raiz2),
        2: (1/raiz2) - 1j*(1/raiz2),
        3: (-1/raiz2) - 1j*(1/raiz2)
    }
    # mapeamento para cada valor no vetor
    symbols = np.array([mapping[v] for v in values])
    return symbols

# Gerar vetor aleatório com N = 1000 valores entre 0 e 3
N = 1000
random_values = np.random.randint(0, 4, N)

# Obter os símbolos modulados
symbols = modulator(random_values)

# Plotar a constelação
plt.figure(figsize=(6,6))
plt.scatter(symbols.real, symbols.imag, s=10, alpha=0.5)
plt.title("Constelação QPSK (Quadrature Phase Shift Keying)")
plt.xlabel("Parte Real")
plt.ylabel("Parte Imaginária")
plt.grid(True )
plt.axhline(0,color = 'black',linewidth=1,linestyle='--')
plt.axvline(0,color = 'black',linewidth=1,linestyle='--')
plt.show()
```

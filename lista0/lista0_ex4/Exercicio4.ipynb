```python
import numpy as np
import matplotlib.pyplot as plt


# Definir a seed para reprodutibilidade
np.random.seed(42)

# o número de elementos
M = 10000

x = np.random.uniform(0, 1, M)
y = np.random.uniform(0, 1, M)

# Calcular z como a soma de x e y
z = x + y

# Plotar o histograma normalizado de z
plt.figure(figsize=(8, 6))
plt.hist(z, bins=50, density=True, alpha=0.7, color='b', edgecolor='black', label="Histograma de z")


plt.xlabel("z")
plt.ylabel("Densidade")
plt.title("Histograma Normalizado de z = x + y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np

# Criando arrays 1D
array1 = np.array([0, 1, 0, 1])
array2 = np.array([0, 0, 1, 1])
```

```python
# Concatenando
matriz_2d = np.c_[array1, array2]
# Dimensão da matriz
print("Dimensão da matriz:", matriz_2d.shape)
```

```python
# Conteúdo da matriz
print("Matriz resultante da concatenação:")
print(matriz_2d)
```

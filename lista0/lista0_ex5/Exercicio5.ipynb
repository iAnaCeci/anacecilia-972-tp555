```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
# vetores x1 e x2 espaçados entre -10 e 11 com passo de 0.25
x1 = np.arange(-10, 11.25, 0.25)
x2 = np.arange(-10, 11.25, 0.25)

#(grid)
X1, X2 = np.meshgrid(x1, x2)

# equação
Y = X1**2 + X2**2

# gráfico 3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plotar a superfície
ax.plot_surface(X1, X2, Y, cmap='viridis')

# Nomear eixos (a, y, z)
ax.set_xlabel('a')
ax.set_ylabel('y')
ax.set_zlabel('z')

# Mostrar gráfico
plt.show()


```

```
import numpy as np
import matplotlib.pyplot as plt
```

```
# Função para gerar dados sintéticos
def generate_data(N=1000, func=lambda X: 2*X[:, 0], noise_std=10.0, D=1):
    X = np.random.rand(N, D) * 10  # entradas aleatórias entre 0 e 10
    y_true = func(X).reshape(-1, 1)
    noise = noise_std * np.random.randn(N, 1)
    y_noisy = y_true + noise
    return X, y_noisy, y_true
```

```
# Equação normal (solução fechada)
def normal_equation(X, y):
    X_b = np.c_[np.ones((X.shape[0], 1)), X]
    theta_opt = np.linalg.pinv(X_b.T @ X_b) @ X_b.T @ y
    return theta_opt
```

```
# Gradiente descendente para múltiplos atributos
def gradient_descent(X, y, alpha=1e-5, max_iter=10000, tol=1e-5):
    N, D = X.shape
    X_b = np.c_[np.ones((N, 1)), X]  # adiciona bias
    theta = np.zeros((D + 1, 1))
    J_hist = []

    for i in range(max_iter):
        y_hat = X_b @ theta
        gradient = -(2/N) * X_b.T @ (y - y_hat)
        theta -= alpha * gradient
        J = (1/N) * np.sum((y - y_hat) ** 2)
        J_hist.append(J)

        if i > 0 and abs(J_hist[-1] - J_hist[-2]) < tol:
            break

    return theta, J_hist
```

```
# Exemplo de uso
# Altere a função aqui conforme necessário
func = lambda X: 2*X[:, 0] + 3  # Pode ser substituída por outra função

X, y_noisy, y_true = generate_data(D=1, func=func)
theta_opt = normal_equation(X, y_noisy)
theta_gd, J_hist = gradient_descent(X, y_noisy)

print("Solução exata (equação normal):", theta_opt.ravel())
print("Solução via gradiente descendente:", theta_gd.ravel())
```

```
# Visualização da função de custo ao longo das iterações
plt.plot(J_hist)
plt.yscale('log')
plt.xlabel('Iterações')
plt.ylabel('Erro quadrático médio (MSE)')
plt.title('Convergência do Gradiente Descendente')
plt.grid()
plt.show()
```

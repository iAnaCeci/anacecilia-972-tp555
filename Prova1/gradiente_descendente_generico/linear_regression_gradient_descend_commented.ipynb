```
# Importação das bibliotecas necessárias
import numpy as np
import matplotlib.pyplot as plt
```

```
# Função para gerar dados sintéticos
def generate_data(N=1000, func=lambda X: 2*X[:, 0], noise_std=10.0, D=1):
    '''
    Gera uma base de dados sintética com ruído.
    - N: número de exemplos
    - func: função alvo aplicada sobre X
    - noise_std: desvio padrão do ruído
    - D: número de atributos
    '''
    X = np.random.rand(N, D) * 10
    y_true = func(X).reshape(-1, 1)
    noise = noise_std * np.random.randn(N, 1)
    y_noisy = y_true + noise
    return X, y_noisy, y_true
```

```
# Função para calcular a solução analítica (equação normal)
def normal_equation(X, y):
    '''
    Resolve a regressão linear com a equação normal.
    Retorna os coeficientes ótimos (incluindo bias).
    '''
    X_b = np.c_[np.ones((X.shape[0], 1)), X]
    theta_opt = np.linalg.pinv(X_b.T @ X_b) @ X_b.T @ y
    return theta_opt
```

```
# Função de gradiente descendente
def gradient_descent(X, y, alpha=0.01, max_iter=10000, tol=1e-5):
    '''
    Executa o gradiente descendente para regressão linear.
    - X: matriz de entrada
    - y: vetor de saída com ruído
    - alpha: taxa de aprendizado
    - max_iter: número máximo de iterações
    - tol: tolerância para convergência
    '''
    N, D = X.shape
    X_b = np.c_[np.ones((N, 1)), X]
    theta = np.zeros((D + 1, 1))
    J_hist = []

    for i in range(max_iter):
        y_hat = X_b @ theta
        gradient = -(2/N) * X_b.T @ (y - y_hat)
        theta -= alpha * gradient
        J = (1/N) * np.sum((y - y_hat)**2)
        J_hist.append(J)
        if i > 0 and abs(J_hist[-1] - J_hist[-2]) < tol:
            break

    return theta, J_hist
```

```
# Exemplo de uso com função linear
func = lambda X: 2 * X[:, 0]

# Gerando dados
X, y_noisy, y_true = generate_data(func=func)

# Aplicando solução analítica e gradiente descendente
theta_opt = normal_equation(X, y_noisy)
theta_gd, J_hist = gradient_descent(X, y_noisy)

print("Solução exata:", theta_opt.ravel())
print("Solução GD:", theta_gd.ravel())
```

```
# Visualização da curva de custo (Erro vs Iteração)
plt.plot(J_hist)
plt.yscale('log')
plt.xlabel('Iterações')
plt.ylabel('Erro quadrático médio (MSE)')
plt.title('Convergência do Gradiente Descendente')
plt.grid()
plt.show()
```

```
# Importações
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
```

```
# Funções de GD: Mini-Batch e Estocástico
def mini_batch_gd(X, y, alpha=0.01, epochs=200, batch_size=32):
    N = X.shape[0]
    theta = np.zeros((X.shape[1], 1))
    J_hist = []

    for epoch in range(epochs):
        indices = np.random.permutation(N)
        X_shuffled = X[indices]
        y_shuffled = y[indices]

        for i in range(0, N, batch_size):
            X_batch = X_shuffled[i:i+batch_size]
            y_batch = y_shuffled[i:i+batch_size]

            y_pred = X_batch @ theta
            error = y_pred - y_batch
            grad = (2 / batch_size) * X_batch.T @ error
            theta -= alpha * grad

        y_pred_all = X @ theta
        mse = np.mean((y_pred_all - y)**2)
        J_hist.append(mse)

    return theta, J_hist

def sgd(X, y, alpha=0.001, epochs=200):
    N = X.shape[0]
    theta = np.zeros((X.shape[1], 1))
    J_hist = []

    for epoch in range(epochs):
        indices = np.random.permutation(N)

        for i in indices:
            xi = X[i:i+1]
            yi = y[i:i+1]

            y_pred = xi @ theta
            error = y_pred - yi
            grad = 2 * xi.T @ error
            theta -= alpha * grad

        y_pred_all = X @ theta
        mse = np.mean((y_pred_all - y)**2)
        J_hist.append(mse)

    return theta, J_hist
```

```
# CASO 1: 1 atributo (x)
N = 1000
x1 = np.random.uniform(-5, 5, (N, 1))
y1 = 3 + 2 * x1 + np.random.normal(0, 1, (N, 1))
X1 = np.c_[np.ones((N, 1)), x1]
```

```
# Mini-Batch GD - caso 1
theta_mini1, J_mini1 = mini_batch_gd(X1, y1, alpha=0.01, batch_size=64)

# SGD - caso 1
theta_sgd1, J_sgd1 = sgd(X1, y1, alpha=0.001)

print("Caso 1 - 1 atributo:")
print("Mini-Batch GD (a0, a1):", theta_mini1.ravel())
print("SGD (a0, a1):", theta_sgd1.ravel())
```

```
# Gráficos do erro - caso 1
plt.figure(figsize=(8,4))
plt.plot(J_mini1, label="Mini-Batch GD")
plt.plot(J_sgd1, label="SGD")
plt.yscale("log")
plt.title("Erro - Caso 1 (1 atributo)")
plt.xlabel("Épocas")
plt.ylabel("MSE")
plt.grid()
plt.legend()
plt.show()
```

```
# CASO 2: 2 atributos (x, x²)
x2 = np.random.uniform(-5, 5, (N, 1))
y2 = 3 + 1.5 * x2 + 2.3 * x2**2 + np.random.normal(0, 1, (N, 1))
X2 = np.c_[np.ones((N, 1)), x2, x2**2]
```

```
# Mini-Batch GD - caso 2
theta_mini2, J_mini2 = mini_batch_gd(X2, y2, alpha=0.01, batch_size=64)

# SGD - caso 2
theta_sgd2, J_sgd2 = sgd(X2, y2, alpha=0.001)

print("Caso 2 - 2 atributos:")
print("Mini-Batch GD (a0, a1, a2):", theta_mini2.ravel())
print("SGD (a0, a1, a2):", theta_sgd2.ravel())
```

```
# Gráficos do erro - caso 2
plt.figure(figsize=(8,4))
plt.plot(J_mini2, label="Mini-Batch GD")
plt.plot(J_sgd2, label="SGD")
plt.yscale("log")
plt.title("Erro - Caso 2 (2 atributos)")
plt.xlabel("Épocas")
plt.ylabel("MSE")
plt.grid()
plt.legend()
plt.show()
```

```
# Importações
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
```

```
# Função de Gradiente Descendente em Batelada
def batch_gradient_descent(X, y, alpha=0.01, max_epochs=1000, tol=1e-6):
    N, D = X.shape
    theta = np.zeros((D, 1))
    J_hist = []

    for epoch in range(max_epochs):
        y_hat = X @ theta
        error = y_hat - y
        grad = (2/N) * X.T @ error
        theta -= alpha * grad
        mse = np.mean(error**2)
        J_hist.append(mse)

        if epoch > 0 and abs(J_hist[-2] - J_hist[-1]) < tol:
            break

    return theta, J_hist
```

```
# CASO 1: Apenas 1 atributo (x)
N = 1000
x1 = np.random.uniform(-5, 5, (N, 1))
y1 = 3 + 2 * x1 + np.random.normal(0, 1, (N, 1))

# Matriz X com bias + x
X1 = np.c_[np.ones((N, 1)), x1]

# Executa o GD
theta1, J_hist1 = batch_gradient_descent(X1, y1)

print("Caso 1 - 1 atributo:")
print("Pesos encontrados (a0, a1):", theta1.ravel())
```

```
# Gráfico do erro - caso 1
plt.figure(figsize=(7,4))
plt.plot(J_hist1)
plt.yscale("log")
plt.title("Convergência do GD - 1 atributo")
plt.xlabel("Épocas")
plt.ylabel("MSE")
plt.grid()
plt.show()
```

```
# CASO 2: Dois atributos (x, x²)
x2 = np.random.uniform(-5, 5, (N, 1))
y2 = 3 + 1.5 * x2 + 2.3 * x2**2 + np.random.normal(0, 1, (N, 1))

# Matriz X com bias + x + x²
X2 = np.c_[np.ones((N, 1)), x2, x2**2]

# Executa o GD
theta2, J_hist2 = batch_gradient_descent(X2, y2)

print("Caso 2 - 2 atributos (x e x²):")
print("Pesos encontrados (a0, a1, a2):", theta2.ravel())
```

```
# Gráfico do erro - caso 2
plt.figure(figsize=(7,4))
plt.plot(J_hist2)
plt.yscale("log")
plt.title("Convergência do GD - 2 atributos (x e x²)")
plt.xlabel("Épocas")
plt.ylabel("MSE")
plt.grid()
plt.show()
```

```
# Comparação visual entre os dois ajustes
x_plot = np.linspace(-5, 5, 200).reshape(-1, 1)

# Caso 1
X_plot1 = np.c_[np.ones_like(x_plot), x_plot]
y_pred1 = X_plot1 @ theta1

# Caso 2
X_plot2 = np.c_[np.ones_like(x_plot), x_plot, x_plot**2]
y_pred2 = X_plot2 @ theta2

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.scatter(x1, y1, alpha=0.2, label="Dados")
plt.plot(x_plot, y_pred1, 'r', label="GD")
plt.title("Ajuste com 1 atributo")
plt.legend()
plt.grid()

plt.subplot(1,2,2)
plt.scatter(x2, y2, alpha=0.2, label="Dados")
plt.plot(x_plot, y_pred2, 'g', label="GD")
plt.title("Ajuste com 2 atributos")
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```

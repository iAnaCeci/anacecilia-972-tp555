```
# Importações
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
```

```
# Geração dos dados com 1 atributo
N = 1000
x = np.random.uniform(-5, 5, (N, 1))  # atributo único
y_true = 3 + 2 * x  # função real
w = np.random.normal(0, 1, (N, 1))  # ruído
y_noisy = y_true + w  # saída com ruído
```

```
# Função solução pela equação normal
def normal_equation(x, y):
    X_b = np.c_[np.ones((len(x), 1)), x]
    theta = np.linalg.pinv(X_b.T @ X_b) @ X_b.T @ y
    return theta
```

```
# Função gradiente descendente para 1 atributo + bias
def gradient_descent(x, y, alpha=0.01, max_iter=1000, tol=1e-6):
    N = len(y)
    X_b = np.c_[np.ones((N, 1)), x]
    theta = np.zeros((2, 1))
    J_hist = []

    for i in range(max_iter):
        y_hat = X_b @ theta
        error = y_hat - y
        grad = (2/N) * X_b.T @ error
        theta -= alpha * grad
        J = np.mean(error**2)
        J_hist.append(J)

        if i > 1 and abs(J_hist[-2] - J_hist[-1]) < tol:
            break

    return theta, J_hist
```

```
# Executando os dois métodos
theta_opt = normal_equation(x, y_noisy)
theta_gd, J_hist = gradient_descent(x, y_noisy)

print("Solução via Equação Normal:", theta_opt.ravel())
print("Solução via Gradiente Descendente:", theta_gd.ravel())
```

```
# Plot do erro
plt.figure(figsize=(8, 4))
plt.plot(J_hist)
plt.yscale("log")
plt.xlabel("Iterações")
plt.ylabel("Erro Quadrático Médio (MSE)")
plt.title("Convergência do Gradiente Descendente")
plt.grid()
plt.show()
```

```
# Comparando visualmente
x_sorted = np.sort(x, axis=0)
X_plot = np.c_[np.ones((len(x_sorted), 1)), x_sorted]
y_pred_normal = X_plot @ theta_opt
y_pred_gd = X_plot @ theta_gd

plt.figure(figsize=(8, 5))
plt.scatter(x, y_noisy, alpha=0.2, label="Dados com ruído")
plt.plot(x_sorted, y_pred_normal, 'r-', label="Equação Normal")
plt.plot(x_sorted, y_pred_gd, 'g--', label="Gradiente Descendente")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.title("Ajuste da Regressão Linear com 1 Atributo")
plt.grid()
plt.show()
```

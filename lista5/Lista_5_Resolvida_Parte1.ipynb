```
# Exercício 4
# Fronteira de decisão para ha(x) = f(5 - x1)

import numpy as np
import matplotlib.pyplot as plt

# Definindo a equação da fronteira: 5 - x1 = 0 => x1 = 5

x1 = np.linspace(0, 10, 100)
x2 = np.linspace(0, 10, 100)
X1, X2 = np.meshgrid(x1, x2)

# Plotando a linha de decisão
plt.figure(figsize=(6,6))
plt.plot([5, 5], [0, 10], 'k--', label='Fronteira de decisão (x1=5)')
plt.fill_betweenx([0,10], 5, 10, color='lightblue', alpha=0.3, label='Classe y=0')
plt.fill_betweenx([0,10], 0, 5, color='salmon', alpha=0.3, label='Classe y=1')

plt.xlim(0, 10)
plt.ylim(0, 10)
plt.xlabel('$x_1$')
plt.ylabel('$x_2$')
plt.title('Fronteira de Decisão')
plt.legend()
plt.grid()
plt.show()
```

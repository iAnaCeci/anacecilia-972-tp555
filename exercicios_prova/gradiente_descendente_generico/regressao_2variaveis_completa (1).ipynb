```
# Importações
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1234)
```

```
# Geração dos dados com 2 variáveis
M = 1000
x1 = np.random.uniform(-5, 5, M)
x2 = np.random.uniform(-5, 5, M)
w = np.random.normal(0, np.sqrt(10), M)

# Função objetivo com interação e quadráticos
y = 3 + 1.5 * x1 + 2.3 * x2 + 0.7 * x1**2 + 0.5 * x2**2 + 1.2 * x1 * x2 + w
```

```
# Matriz de atributos: bias, x1, x2, x1², x2², x1*x2
X = np.c_[
    np.ones((M, 1)),
    x1.reshape(-1, 1),
    x2.reshape(-1, 1),
    (x1**2).reshape(-1, 1),
    (x2**2).reshape(-1, 1),
    (x1 * x2).reshape(-1, 1)
]
```

```
# Equação normal
a_normal = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)

# Predição
yhat_normal = X.dot(a_normal)

# MSE
mse_normal = (1.0/M)*np.sum((y - yhat_normal)**2)

print("Solução (Equação Normal):")
for i, a in enumerate(a_normal):
    print(f"a{i} = {a:.4f}")
print(f"MSE: {mse_normal:.6f}")
```

```
# Gradiente descendente em batelada
def batchGradientDescent(X, y, alpha=0.00001, max_epochs=1000):
    M = len(y)
    y = y.reshape(M, 1)
    a = np.random.randn(X.shape[1], 1) * 0.1
    J_hist = []
    for epoch in range(max_epochs):
        y_hat = X @ a
        error = y_hat - y
        grad = (2/M) * X.T @ error
        a -= alpha * grad
        mse = np.mean(error**2)
        J_hist.append(mse)
        if epoch > 1 and abs(J_hist[-1] - J_hist[-2]) < 1e-6:
            break
    return a, J_hist
```

```
# Executa o gradiente descendente
a_gd, errors_gd = batchGradientDescent(X, y, alpha=0.00001, max_epochs=1000)

# Predição final
yhat_gd = X @ a_gd
mse_gd = np.mean((y - yhat_gd.ravel())**2)

print("Solução (Gradiente Descendente):")
for i, a in enumerate(a_gd):
    print(f"a{i} = {a[0]:.4f}")
print(f"MSE: {mse_gd:.6f}")
```

```
# Gráfico do erro
plt.figure(figsize=(8, 4))
plt.plot(errors_gd)
plt.yscale("log")
plt.xlabel("Épocas")
plt.ylabel("Erro Quadrático Médio (MSE)")
plt.title("Convergência do Gradiente Descendente")
plt.grid()
plt.show()
```
